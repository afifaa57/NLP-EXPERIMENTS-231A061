## Step 1: Text Preprocessing

First, I'll install necessary libraries (NLTK and spaCy) and download their data. Then, I'll define an input text, tokenize it, and perform POS tagging.

```python
print("Name:QURESHI AFIFA ARIF UIN:231A061")
import nltk
import spacy

# Download NLTK data
print("Downloading NLTK resources...")
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng') # Added to resolve LookupError for averaged_perceptron_tagger_eng
nltk.download('maxent_ne_chunker')
nltk.download('words')
print("NLTK resources downloaded.")

# Download spaCy English model (if not already downloaded)
try:
    spacy.load('en_core_web_sm')
except OSError:
    print('Downloading spaCy model "en_core_web_sm"...')
    !python -m spacy download en_core_web_sm
    print('SpaCy model downloaded.')
```

```python
print("Name:QURESHI AFIFA ARIF UIN:231A061")
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag

# Input Text Corpus
input_text = "John visited the famous Tower of London in 2023. He bought a souvenir from a shop near Buckingham Palace."
print(f"Input Text: {input_text}")

# Tokenization
tokens = word_tokenize(input_text)
print(f"\nTokens: {tokens}")

# POS Tagging
pos_tags = pos_tag(tokens)
print(f"\nPOS Tags: {pos_tags}")
```

## Step 2: Chunking to Extract Phrases

Next, I'll define a chunk pattern to extract noun phrases using NLTK's `RegexpParser` and apply it to the POS-tagged tokens.

```python
print("Name:QURESHI AFIFA ARIF UIN:231A061")
from nltk.chunk import RegexpParser

# Define Chunk Patterns (for Noun Phrases - NP)
# NP: {<DT>?<JJ>*<NNP>+} - Optional Determiner, zero or more Adjectives, one or more Proper Nouns
# NP: {<DT>?<JJ>*<NN>} - Optional Determiner, zero or more Adjectives, a Noun

# Let's use a more general NP pattern for this example, including proper nouns and common nouns.
# NP: {<DT>?<JJ.*>*<NN.*>+}
chunk_pattern = r"""NP: {<DT>?<JJ.*>*<NN.*>+}"""

# Create a chunk parser
chunk_parser = RegexpParser(chunk_pattern)

# Apply Chunking
tree = chunk_parser.parse(pos_tags)

# Extract and display chunks
print("Extracted Chunks (Noun Phrases):")
for subtree in tree.subtrees():
    if subtree.label() == 'NP':
        print(subtree)
```

## Step 3 & 4: Named Entity Recognition (NER) and Extraction

Now, I'll use a pre-trained spaCy model (`en_core_web_sm`) to perform Named Entity Recognition on the input text and extract the identified entities.

```python
print("Name:QURESHI AFIFA ARIF UIN:231A061")
import spacy

# Load pre-trained spaCy model
nlp = spacy.load('en_core_web_sm')

# Apply NER to the input text
doc = nlp(input_text)

# Extract Named Entities
print("Named Entities:")
entities = []
for ent in doc.ents:
    entities.append((ent.text, ent.label_))
    print(f"- {ent.text} ({ent.label_})")

print(f"\nOutput: {entities}")
```

## Step 5: Evaluation and Comparison

For a full evaluation, you would compare these results against a manually annotated dataset. For this example, we can manually verify the outputs.

```python
# Manual verification of extracted chunks and entities
print("Name:QURESHI AFIFA ARIF UIN:231A061")
print(f"Input Sentence: {input_text}")
print(f"\nPOS Tags: {pos_tags}")
print("\nManually verify Noun Phrases from Chunking:")
# Expected Noun Phrases based on the pattern might include: 'John', 'the famous Tower of London', 'a souvenir', 'a shop', 'Buckingham Palace'
# Note: NLTK's RegexpParser with simple patterns might not catch all nuances compared to statistical parsers.
chunk_parser = RegexpParser(chunk_pattern)
tree = chunk_parser.parse(pos_tags)
for subtree in tree.subtrees():
    if subtree.label() == 'NP':
        print(subtree.leaves())

print("\nManually verify Named Entities from spaCy:")
# Expected Entities based on spaCy: [('John', 'PERSON'), ('Tower of London', 'LOC'), ('2023', 'DATE'), ('Buckingham Palace', 'LOC')]
for ent in doc.ents:
    print(f"- {ent.text} ({ent.label_})")

print("\nComparison with Example Output from Prompt:")
print("Prompt Example Output: [('John', 'PERSON'), ('Tower of London', 'LOCATION'), ('2023', 'DATE')] ")
print("My Output: ", entities)
print("The outputs are largely consistent, with spaCy providing additional entities from the extended text.")
```
