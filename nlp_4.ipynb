```python
print("Name:QURESHI AFIFA ARIF UIN:231A061")
import nltk
nltk.download('punkt_tab')
tokens = word_tokenize(text_lower)
print(tokens)
```

```python
print("Name:QURESHI AFIFA ARIF UIN:231A061")
text_clean = []
for word in tokens:
      if word not in string.punctuation:
           text_clean.append(word)
builtins.print(text_clean)
```

```python
print("Name:QURESHI AFIFA ARIF UIN:231A061")
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(tok , pos='v') for tok in text_clean]
builtins.print(lemmas)
```

```python
print("Name:QURESHI AFIFA ARIF UIN:231A061")
lemmatizer.lemmatize("skills", pos='v')
```

```python
[(token, lemma) for token, lemma in zip(tokens, lemmas)]
```

```python
morphological_info = {}
for token, lema in zip(tokens, lemmas):
    if token != lema:
        if token.endswith('ing'):
            analysis = f"Inflectional: Present participle/continuous form of '{lema}'"
        elif token.endswith("er") or token.endswith("ation"):
            analysis = f"Derivational: '{lema}' + -er -> person who does the action"
        else:
            analysis = f"Inflectional or irregular form of '{lema}'"
    else:
        analysis = "Base/root form"
    morphological_info[token] = analysis
```
